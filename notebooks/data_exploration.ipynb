```python

import pandas as pd
import numpy as np

# Start with the original dataset
cleaned = df.copy()

print("Original number of rows:", len(cleaned))

# 1. Convert date and time columns
cleaned["tpep_pickup_datetime"] = pd.to_datetime(
    cleaned["tpep_pickup_datetime"], errors="coerce"
)
cleaned["tpep_dropoff_datetime"] = pd.to_datetime(
    cleaned["tpep_dropoff_datetime"], errors="coerce"
)

# 2. Remove rows missing essential values
required = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime",
    "trip_distance",
    "fare_amount",
    "pickup_longitude",
    "pickup_latitude",
    "dropoff_longitude",
    "dropoff_latitude"
]

cleaned = cleaned.dropna(subset=required).copy()
print("After removing missing values:", len(cleaned))

# 3. Remove invalid times, distances and fares
cleaned = cleaned[
    (cleaned["tpep_dropoff_datetime"] >
     cleaned["tpep_pickup_datetime"]) &
    (cleaned["trip_distance"] > 0) &
    (cleaned["fare_amount"] > 0)
].copy()

print("After removing invalid trips:", len(cleaned))

# 4. Keep trips within approximate NYC-area coordinates
cleaned = cleaned[
    cleaned["pickup_longitude"].between(-74.3, -73.6) &
    cleaned["pickup_latitude"].between(40.4, 41.0) &
    cleaned["dropoff_longitude"].between(-74.3, -73.6) &
    cleaned["dropoff_latitude"].between(40.4, 41.0)
].copy()

print("After geographical filtering:", len(cleaned))

# 5. Calculate trip duration in minutes
cleaned["duration_min"] = (
    cleaned["tpep_dropoff_datetime"] -
    cleaned["tpep_pickup_datetime"]
).dt.total_seconds() / 60

# 6. Remove implausible durations, distances and fares
cleaned = cleaned[
    cleaned["duration_min"].between(1, 180) &
    cleaned["trip_distance"].between(0.1, 100) &
    cleaned["fare_amount"].between(2.5, 300)
].copy()

# 7. Remove duplicate rows
cleaned = cleaned.drop_duplicates().reset_index(drop=True)

# 8. Create useful features
cleaned["hour"] = cleaned["tpep_pickup_datetime"].dt.hour
cleaned["day_of_week"] = cleaned["tpep_pickup_datetime"].dt.dayofweek
cleaned["day"] = cleaned["tpep_pickup_datetime"].dt.day

cleaned["pickup_cell_lat"] = cleaned["pickup_latitude"].round(3)
cleaned["pickup_cell_lon"] = cleaned["pickup_longitude"].round(3)

# 9. Display final results
print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned))
print("Rows removed:", len(df) - len(cleaned))
print("Columns after feature engineering:", len(cleaned.columns))

display(cleaned.head())

# 10. Save only the cleaned dataset
cleaned.to_csv("cleaned_trips.csv", index=False)

print("\nCleaned dataset saved as cleaned_trips.csv")
```

```python
# Arrange trips in chronological order
cleaned = cleaned.sort_values("tpep_pickup_datetime").reset_index(drop=True)

# Split 80% for training and 20% for testing
split_index = int(len(cleaned) * 0.8)

train_trips = cleaned.iloc[:split_index].copy()
test_trips = cleaned.iloc[split_index:].copy()

# Display the results
print("Total cleaned trips:", len(cleaned))
print("Training trips (80%):", len(train_trips))
print("Testing trips (20%):", len(test_trips))

print("\nTraining data date range:")
print(train_trips["tpep_pickup_datetime"].min())
print("to")
print(train_trips["tpep_pickup_datetime"].max())

print("\nTesting data date range:")
print(test_trips["tpep_pickup_datetime"].min())
print("to")
print(test_trips["tpep_pickup_datetime"].max())
```

```python
# Create pickup counts and high-demand labels

# Create the start of each pickup hour
cleaned["pickup_hour"] = (cleaned["tpep_pickup_datetime"].dt.floor("h"))

# Count pickups per area per hour
hourly_demand = (cleaned.groupby(["pickup_cell_lat", "pickup_cell_lon", "pickup_hour"]).size().reset_index(name="pickup_count"))

# Find the threshold using training-period observations only
train_end = train_trips["tpep_pickup_datetime"].max()

training_demand = hourly_demand[hourly_demand["pickup_hour"] <= train_end.floor("h")]
demand_threshold = training_demand["pickup_count"].quantile(0.80)

# Label high-demand area-hours
hourly_demand["high_demand"] = (hourly_demand["pickup_count"] > demand_threshold).astype(int)

print("Area-hour records:", len(hourly_demand))
print("80th percentile threshold:", demand_threshold)
print("\nLabel distribution:")
print(hourly_demand["high_demand"].value_counts())
print("\nSample records:")
display(hourly_demand.head())
```

```python
# Rebuild demand labels using area and hour of day

# Count trips for each pickup area and hour of day
area_hour_demand = (cleaned.groupby(["pickup_cell_lat", "pickup_cell_lon", "hour"]).size().reset_index(name="pickup_count"))

# Calculate threshold from the distribution of area-hour counts
demand_threshold = area_hour_demand["pickup_count"].quantile(0.80)

# Label the busiest 20% approximately
area_hour_demand["high_demand"] = (area_hour_demand["pickup_count"] > demand_threshold).astype(int)

print("Area-hour combinations:", len(area_hour_demand))
print("80th percentile threshold:", demand_threshold)
print("\nLabel distribution:")
print(area_hour_demand["high_demand"].value_counts())
print("\nLabel percentages:")
print((area_hour_demand["high_demand"].value_counts(normalize=True) * 100).round(2))

display(area_hour_demand.head())
```

```python
# K-Means clustering for pickup hotspots
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Use pickup coordinates
locations = cleaned[["pickup_latitude", "pickup_longitude"]].dropna()

# Sample up to 5000 locations to keep processing fast
locations_sample = locations.sample(n=min(5000, len(locations)),random_state=42)

# Test different numbers of clusters
k_values = range(2, 10)
inertias = []
silhouette_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = model.fit_predict(locations_sample)

    inertias.append(model.inertia_)
    silhouette_scores.append(
        silhouette_score(locations_sample, labels)
    )
```

```python
# Plot elbow curve
plt.figure(figsize=(8, 5))
plt.plot(list(k_values), inertias, marker="o")
plt.title("Elbow Method for Pickup Hotspots")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()

# Plot silhouette scores
plt.figure(figsize=(8, 5))
plt.plot(list(k_values), silhouette_scores, marker="o")
plt.title("Silhouette Score for Pickup Hotspots")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.grid(True)
plt.show()
```

```python
# Visualize pickup hotspots using Matplotlib
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Prepare pickup coordinates
locations = cleaned[["pickup_latitude", "pickup_longitude"]].dropna()

# Sample data for faster plotting
locations_sample = locations.sample(
    n=min(5000, len(locations)),
    random_state=42)

# Group pickup locations into 5 clusters
kmeans = KMeans(n_clusters=5,random_state=42,n_init=10)

cluster_labels = kmeans.fit_predict(locations_sample)

# Plot clusters
plt.figure(figsize=(10, 7))
plt.scatter(locations_sample["pickup_longitude"],locations_sample["pickup_latitude"],c=cluster_labels,cmap="viridis",s=8,alpha=0.6)

plt.title("NYC Taxi Pickup Hotspot Clusters")
plt.xlabel("Pickup Longitude")
plt.ylabel("Pickup Latitude")
plt.grid(True, alpha=0.3)
plt.show()

# Display cluster centers
print("Cluster centers (latitude, longitude):")
print(kmeans.cluster_centers_[:, [1, 0]])
```
